```python
def find_motive_dfs(obj, X, coefs, monoms_H, monoms_X, n_rounds):
    def dfs(candidate, big_part, n_monom, n_round):
        if n_monom == len(monoms_H):
            m = find_motive_low(candidate, X)
            if m is not None:
                return (big_part + m).expand()
            return None
        monom = monoms_H[n_monom]
        new_n_monom = n_monom + 1 if n_round==n_rounds-1 else n_monom
        new_n_round = 0 if n_round==n_rounds-1 else n_round+1
        for coef in coefs:
            m = (candidate-coef*monom).expand()
            if any(term.could_extract_minus_sign() for term in m.as_ordered_terms()):
                continue
            result = dfs(m, big_part+coef*monoms_X[n_monom], new_n_monom, new_n_round)
            if result is not None:
                return result
        return None
    res = dfs(obj, 0, 0, 0)
    return [res] if res is not None else []
```

```python
t0 = time()
coefs = [0] + [L**i for i in range(9)]
monoms_H = []
monoms_X = []
n_rounds = 2
motives = find_motive_bfs(motive_H_clean, X, coefs, monoms_H, monoms_X, n_rounds)
print(time()-t0)
```
