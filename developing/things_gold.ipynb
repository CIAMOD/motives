```python
from tqdm import tqdm
```

```python
g = 2
r = 4
d = 1 # dejar así
X = Curve("X", g)
H = X.curve_chow
J = Jacobian(X).get_lambda_var(1)
```

```python
motive_H = (motive_generic(X, r, d)/J).expand().simplify()
motive_H_clean = motive_generic_clean(X, r, d)
print(motive_H)
```

```python
coefs = [0] + [L**i for i in range(12)]
```

```python
candidates = {(motive_H_clean, 0)}
monoms = [sym_lambda_in_chow(X, 3)*sym_lambda_in_chow(X, 1), sym_lambda_in_chow(X, 3)]
monoms_X = [sym_lambda(X, 3)*sym_lambda(X, 1), sym_lambda(X, 3)]
for n, monom in enumerate(monoms):
    for i in range(6):
        print(f"{n+1} monomial, {i+1}th round, {len(candidates)} candidates")
        new_candidates = set()
        for candidate, big_part in tqdm(candidates):
            for coef in coefs:
                m = (candidate - coef*monom).expand()
                if not any(term.could_extract_minus_sign() for term in m.as_ordered_terms()):
                    new_candidates.add((m, big_part+coef*monoms_X[n]))
        candidates = new_candidates

len(candidates)
```

```python
porfa = set()
for candidate, big_part in tqdm(candidates):
    m = find_motive_low(candidate, X)
    if m != None:
        porfa.add((m, big_part))
```

```python
motives = [m + big_part for m, big_part in porfa]
```

```python
motive = motives[10]
compare(subs_H(motive, X), motive_H)
```
